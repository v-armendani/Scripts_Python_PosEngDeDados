```python
l = list(range(15))
l2 = list(range(0, 15))
l3 = list(range(0, 15, 2)) # Range SSS: start, stop, step
print('Lista range(15): ', l)
print('Lista range(0, 15): ', l2)
print('Lista range(0, 15, 2): ', l3)

l.append(15)
print('Lista com um append(15): ', l)

l.insert(0,6) # no índice '0' inclua o valor '6'
print('Lista com insert(0, 6): ', l)

l.reverse()
print('Lista com reverse(): ', l)

l[0] = 300
print('Lista com reverse(): ', l)
```

```python
contagem = l2.count(2)
print('Lista com count (2): ', contagem) # quantas vezes o número 2 aparece na lista

tamanho = len(l)
print('Usamos len() para encontrar o tamanho da lista: ', tamanho)

l.sort()
l.sort(reverse=True) # ordem reversa --> sort(reverse=True)
print(l)
```

```python
# criar a função ordena()

def ordena(l):
  return len(l)


l = ['python', 'jana', 'C++', 'C']

# l.sort(key=ordena)
l.sort(reverse=True, key=ordena) # ordem reversa --> sort(reverse=True)
print(l)
```

```python
def ordenaDic(dic):
  return dic['Ano']

dic = [
  {'Filme': 'Esqueceram de mim 3', 'Ano': 1998},
  {'Filme': 'Matrix 1', 'Ano': 1999},
  {'Filme': 'Velozes e Furiosos 4', 'Ano': 2009},
  {'Filme': 'Bohemian Rhapsody', 'Ano': 2018}
]

# dic.sort(key=ordenaDic)
dic.sort(reverse=True, key=ordenaDic)

dic
```

## Slice

```python
lista = list(range(20))

print(lista[:5]) # só os 5 primeiros
print(lista[-5:]) # só os 5 últimos *** pq não foi até o 20? (range de 20)
print(lista[:-5]) # retirar os 5 últimos
print(lista)
```

```python

lista[0:20:2] # A notação [0:20:2] significa, então, que queremos uma sublista do 1º ao 20º elemento, pulando de 2 em 2 elementos.
lista[2:-5:2] # A notação [2:-5:2] significa que iremos pular de 2 em 2 elementos e excluir os últimos 5 elementos....iexclui o 15, 16, 17, 18 e 19
```
