```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np

# Carregar dados: duas categorias apenas (binária)
categorias = ['sci.space', 'rec.sport.hockey']
dados = fetch_20newsgroups(subset='all', categories=categorias, remove=('headers', 'footers', 'quotes'))

# Separar textos e rótulos
textos = dados.data
rotulos = dados.target  # 0 = sci.space, 1 = hockey

# Separar treino e teste
X_train_raw, X_test_raw, y_train, y_test = train_test_split(textos, rotulos, test_size=0.2, random_state=42)
print("# Separar treino e teste", X_train_raw[0])

# Tokenização
tokenizer = Tokenizer(num_words=5000, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train_raw)
X_token = tokenizer.fit_on_texts(X_train_raw)

# Converter textos para sequências numéricas
X_train_seq = tokenizer.texts_to_sequences(X_train_raw)
X_test_seq = tokenizer.texts_to_sequences(X_test_raw)
print("# Converter textos para sequências numéricas", X_train_seq[0])

# Padronizar comprimento das sequências
maxlen = 200
X_train = pad_sequences(X_train_seq, maxlen=maxlen, padding='post')
X_test = pad_sequences(X_test_seq, maxlen=maxlen, padding='post')
print("# Padronizar comprimento das sequências", X_train[0])

# Construir o modelo LSTM simples
model = Sequential()
model.add(Embedding(input_dim=5000, output_dim=32, input_length=maxlen))
model.add(LSTM(32))
model.add(Dense(1, activation='sigmoid'))

# Compilar e treinar
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.2)
# print(X_train[0])

# Avaliação
loss, acc = model.evaluate(X_test, y_test)
print(f'\nAcurácia no teste: {acc*100:.2f}%')
```
