2. A Ausência Absoluta de Relação Linear
O coeficiente de correlação de Pearson varia de -1 a 1.

Valores acima de 0.5 (ou abaixo de -0.5) indicam forte relação.

Valores como 0.001 e -0.001 são o equivalente estatístico a zero.
Isso significa que matematicamente, saber a idade (Age), o salário (MonthlyIncome) ou qualquer outro dado dessa tabela não ajuda em absolutamente nada a prever o Target por vias lineares.

3. Por que algumas deram NaN?
As colunas EmployeeCount e StandardHours retornaram NaN (Not a Number). Isso acontece na fórmula de Pearson quando a variância da coluna é zero (ou seja, todos os funcionários têm exatamente o mesmo valor nessa coluna). Como a fórmula exige dividir algo pelo desvio padrão, ocorre uma divisão por zero. O diagnóstico? Elas são completamente inúteis e devem ser excluídas. O mesmo vale para o EmployeeNumber, que é só um identificador.

````python
from scipy.stats import ttest_ind, chi2_contingency

print("--- TESTES ESTATÍSTICOS DE HIPÓTESES ---\n")

# ==========================================
# 1. TESTE T (Variável Numérica vs Alvo)
# Hipótese Nula (H0): A média salarial de quem sai é IGUAL à de quem fica.
# ==========================================
coluna_num = 'MonthlyIncome'
grupo_fica = df[df['Attrition'] == 'No'][coluna_num]
grupo_sai = df[df['Attrition'] == 'Yes'][coluna_num]

stat_t, p_valor_t = ttest_ind(grupo_fica, grupo_sai, equal_var=False)

print(f"1. Teste T de Student para '{coluna_num}'")
print(f"   Média de quem fica: {grupo_fica.mean():.2f}")
print(f"   Média de quem sai:  {grupo_sai.mean():.2f}")
print(f"   P-valor: {p_valor_t:.5f}")

if p_valor_t > 0.05:
    print("   CONCLUSÃO: Aceitamos H0. Não há diferença estatística significativa. O fator é aleatório.\n")
else:
    print("   CONCLUSÃO: Rejeitamos H0. Há uma diferença estatística válida.\n")


# ==========================================
# 2. TESTE QUI-QUADRADO (Variável Categórica vs Alvo)
# Hipótese Nula (H0): Fazer horas extras (OverTime) não tem relação com a saída (Attrition).
# ==========================================
coluna_cat = 'OverTime'
tabela_contingencia = pd.crosstab(df[coluna_cat], df['Attrition'])

chi2_stat, p_valor_chi, dof, expected = chi2_contingency(tabela_contingencia)

print(f"2. Teste Qui-Quadrado para '{coluna_cat}'")
print(f"   P-valor: {p_valor_chi:.5f}")

if p_valor_chi > 0.05:
    print("   CONCLUSÃO: Aceitamos H0. As variáveis são independentes. O fator é aleatório.")
else:
    print("   CONCLUSÃO: Rejeitamos H0. Existe dependência entre as variáveis.")
````

````python
import numpy as np

# 1. Rácios Financeiros (Ganha bem para a idade que tem?)
df['Income_to_Age_Ratio'] = df['MonthlyIncome'] / df['Age']

# 2. Rácios de Fidelidade (Quão estagnado ele está?)
# Usamos +1 no denominador para evitar divisão por zero
df['Company_to_TotalYears'] = df['YearsAtCompany'] / (df['TotalWorkingYears'] + 1)
df['Role_to_CompanyYears'] = df['YearsInCurrentRole'] / (df['YearsAtCompany'] + 1)

# 3. Índice de Desgaste (Juntando distância, horas extras e viagens)
# Transformando as categóricas temporariamente em números para o cálculo
overtime_num = df['OverTime'].map({'Yes': 1, 'No': 0})
travel_num = df['BusinessTravel'].map({'Non-Travel': 0, 'Travel_Rarely': 1, 'Travel_Frequently': 2})

df['Burnout_Index'] = (df['DistanceFromHome'] * 0.2) + (overtime_num * 5) + (travel_num * 2)

# 4. Estabilidade de Liderança
df['Manager_Stability'] = df['YearsWithCurrManager'] / (df['YearsAtCompany'] + 1)
````

````python
import pandas as pd
import numpy as np
import optuna
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.metrics import classification_report, roc_auc_score, roc_curve
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

print("=== 1. FEATURE ENGINEERING ===")
# Criação das variáveis de negócio baseadas em conhecimento de RH
df['Income_to_Age_Ratio'] = df['MonthlyIncome'] / df['Age']
df['Company_to_TotalYears'] = df['YearsAtCompany'] / (df['TotalWorkingYears'] + 1)
df['Role_to_CompanyYears'] = df['YearsInCurrentRole'] / (df['YearsAtCompany'] + 1)
df['Manager_Stability'] = df['YearsWithCurrManager'] / (df['YearsAtCompany'] + 1)

# Variável Alvo Temporária (Numérica)
df['Target'] = df['Attrition'].map({'Yes': 1, 'No': 0})

print("=== 2. PRÉ-PROCESSAMENTO ===")
# Remoção de colunas inúteis e vazamento de dados
cols_to_drop = ['EmployeeCount', 'StandardHours', 'EmployeeNumber', 'Attrition', 'Over18']
df_clean = df.drop(columns=cols_to_drop, errors='ignore')

# Separar Features (X) e Target (y)
X = df_clean.drop(columns=['Target'])
y = df_clean['Target']

# One-Hot Encoding
X_encoded = pd.get_dummies(X, drop_first=True)

# Divisão Treino e Teste (Estratificada para manter os 16% de atrito)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42, stratify=y
)

print("=== 3. CONFIGURAÇÃO DE BALANCEAMENTO (PESOS) ===")
# Calcula o peso da classe minoritária para o XGBoost penalizar os erros
peso_desbalanceamento = y_train.value_counts()[0] / y_train.value_counts()[1]
print(f"Peso aplicado na classe minoritária: {peso_desbalanceamento:.2f}")

print("\n=== 4. OTIMIZAÇÃO DE HIPERPARÂMETROS (OPTUNA) ===")
def objective_xgb(trial):
    # O Optuna vai testar valores aleatórios inteligentes dentro destas faixas
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 200),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'scale_pos_weight': peso_desbalanceamento, # Aplicando o balanceamento aqui!
        'random_state': 42,
        'eval_metric': 'logloss',
        'n_jobs': -1
    }
    
    # Validação cruzada (Cross-Validation) com 3 cortes para garantir que o modelo é robusto
    cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    model = XGBClassifier(**params)
    
    # Usando ROC AUC como métrica guia para o Optuna
    scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='roc_auc', n_jobs=-1)
    return scores.mean()

# Cria o estudo e roda (Ajustado para 15 tentativas (trials) para não demorar muito)
optuna.logging.set_verbosity(optuna.logging.WARNING) # Esconde mensagens excessivas
study_xgb = optuna.create_study(direction='maximize', study_name='XGBoost_Optimization')
study_xgb.optimize(objective_xgb, n_trials=15, show_progress_bar=True)

print(f"\nMelhor Score AUC (Treino/CV): {study_xgb.best_value:.4f}")
print("Melhores Hiperparâmetros encontrados:")
for key, value in study_xgb.best_params.items():
    print(f"  {key}: {value}")

print("\n=== 5. TREINAMENTO FINAL E AVALIAÇÃO (CONJUNTO DE TESTE) ===")
# Pega os melhores parâmetros e treina o modelo final
best_params = study_xgb.best_params
best_params['scale_pos_weight'] = peso_desbalanceamento
best_params['random_state'] = 42

best_xgb = XGBClassifier(**best_params)
best_xgb.fit(X_train, y_train)

# Previsões
y_pred = best_xgb.predict(X_test)
y_prob = best_xgb.predict_proba(X_test)[:, 1]

# Métricas Finais
auc_final = roc_auc_score(y_test, y_prob)
print(f"ROC AUC (Teste): {auc_final:.4f}\n")
print("Relatório de Classificação Final:")
print(classification_report(y_test, y_pred))
````

## Nested Cross-Validation com Optuna e XGBoost

"Para assegurar a máxima robustez metodológica e isolar a estimativa de performance de qualquer viés de otimização de hiperparâmetros (Data Leakage), o pipeline final culminou na implementação de uma Validação Cruzada Aninhada (Nested Cross-Validation). >
Com o Optuna a atuar no loop interno e o modelo sendo testado em partições estritamente isoladas no loop externo, o XGBoost obteve um ROC AUC médio de 0.4997, acompanhado de um desvio padrão de apenas 0.0012. A extrema consistência desta métrica, aliada ao seu valor ancorado na linha de base aleatória (0.50), consolida a tese central deste estudo.

Esgotado o estado-da-arte em engenharia de features, balanceamento de classes e otimização bayesiana robusta, atesta-se que a base de dados atinge o seu Erro Irredutível (Irreducible Error). Conclui-se, de forma irrefutável, que os dados sintéticos analisados possuem entropia máxima em relação ao Target, não apresentando qualquer covariância lógica subjacente que permita o mapeamento de fronteiras de decisão por algoritmos de Machine Learning."

````python
import numpy as np
import pandas as pd
import optuna
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import roc_auc_score
from xgboost import XGBClassifier
import warnings
warnings.filterwarnings('ignore')

print("=== NESTED CROSS-VALIDATION (VALIDAÇÃO CRUZADA ANINHADA) ===\n")

# Relembrando: X_encoded e y já foram criados no script de pré-processamento anterior
# Vamos usar o peso_desbalanceamento que já calculámos (aprox. 5.24)
peso_desbal = y.value_counts()[0] / y.value_counts()[1]

# 1. Configurar o Loop Externo (Avaliação real do modelo)
outer_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
outer_scores = []

# Iterar sobre os cortes do Loop Externo
for fold, (train_idx, test_idx) in enumerate(outer_cv.split(X_encoded, y)):
    print(f"--- Iniciando Fold Externo {fold + 1}/3 ---")
    
    # Separar os dados deste fold
    X_train_outer, X_test_outer = X_encoded.iloc[train_idx], X_encoded.iloc[test_idx]
    y_train_outer, y_test_outer = y.iloc[train_idx], y.iloc[test_idx]
    
    # 2. Configurar o Loop Interno (Onde o Optuna trabalha)
    def objective_inner(trial):
        params = {
            'n_estimators': trial.suggest_int('n_estimators', 50, 150),
            'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.2, log=True),
            'max_depth': trial.suggest_int('max_depth', 3, 6),
            'subsample': trial.suggest_float('subsample', 0.6, 1.0),
            'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
            'scale_pos_weight': peso_desbal,
            'random_state': 42,
            'eval_metric': 'logloss',
            'n_jobs': -1
        }
        
        # Loop Interno: O Optuna faz o seu próprio Cross-Validation apenas nos dados de treino externos
        inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
        model_inner = XGBClassifier(**params)
        
        scores_inner = cross_val_score(model_inner, X_train_outer, y_train_outer, 
                                       cv=inner_cv, scoring='roc_auc', n_jobs=-1)
        return scores_inner.mean()
    
    # Executar o Optuna para este Fold Externo
    optuna.logging.set_verbosity(optuna.logging.WARNING)
    study = optuna.create_study(direction='maximize')
    
    print("Otimizando hiperparâmetros no Loop Interno (Optuna)...")
    study.optimize(objective_inner, n_trials=10, show_progress_bar=False)
    
    # 3. Treinar o Melhor Modelo no Fold Externo Inteiro e Testar
    best_params = study.best_params
    best_params['scale_pos_weight'] = peso_desbal
    best_params['random_state'] = 42
    
    best_model_outer = XGBClassifier(**best_params)
    best_model_outer.fit(X_train_outer, y_train_outer)
    
    # Prever no conjunto de teste que o Optuna NUNCA viu
    y_prob_outer = best_model_outer.predict_proba(X_test_outer)[:, 1]
    auc_outer = roc_auc_score(y_test_outer, y_prob_outer)
    
    outer_scores.append(auc_outer)
    print(f"ROC AUC do Fold Externo {fold + 1}: {auc_outer:.4f}\n")

# 4. Resultado Final do Nested CV
print("=== RESULTADO FINAL (NESTED CV) ===")
print(f"ROC AUC Médio: {np.mean(outer_scores):.4f}")
print(f"Desvio Padrão: {np.std(outer_scores):.4f}")
````
